## Program — Hill climbing for the 4-Queens problem

**Theory.** The 4-Queens task places four queens so no pair shares a row, column or diagonal. The notebook represents one queen per column and a row per list entry. Hill climbing repeatedly chooses a lower-conflict neighboring state; it can stop at a local minimum.

**Objective.** Display the conflict cost of a user-supplied board and each neighbor until a solution or local minimum is reached.

**Libraries.** Python standard library only.

**Algorithm.**
1. Read a permutation of rows. 2. Count attacking pairs. 3. Generate neighbors by swapping two columns’ row values. 4. Move to the first best lower-cost neighbor. 5. Stop at cost 0 or when no neighbor improves the state.

**Complexity.** O(n^3) time per iteration for O(n²) swap neighbors with O(n²) cost evaluation; O(n) state space.

The code below keeps `input()` so you can provide your own values. The final lines run a sample interaction through the same prompts; edit `SAMPLE_INPUTS` to try another case, or call the main function directly for a live terminal run.

**Expected output and explanation.** The output reports the computed result for the entered values.


```python
from contextlib import contextmanager
import builtins

@contextmanager
def sample_input(values):
    original_input = builtins.input
    remaining = iter(map(str, values))
    def demo_input(prompt=''):
        print(prompt, end='')
        try:
            value = next(remaining)
        except StopIteration as exc:
            raise RuntimeError('Add more values to SAMPLE_INPUTS for this run.') from exc
        print(value)
        return value
    builtins.input = demo_input
    try:
        yield
    finally:
        builtins.input = original_input

def cost(state):
    return sum(state[i]==state[j] or abs(state[i]-state[j])==j-i
               for i in range(len(state)) for j in range(i+1,len(state)))
def hill_climb(state):
    step=0
    while True:
        current=cost(state); print(f'\nStep {step}: {state}; cost={current}')
        if current==0: print('Goal reached!'); return state
        best_state=state[:]; best_cost=current
        for i in range(len(state)):
            for j in range(i+1,len(state)):
                neighbor=state[:]; neighbor[i],neighbor[j]=neighbor[j],neighbor[i]
                value=cost(neighbor); print(f'Swap ({i},{j}): {neighbor}; cost={value}')
                if value<best_cost: best_state,best_cost=neighbor,value
        if best_cost>=current: print('Local minimum reached.'); return state
        state=best_state; step+=1
def main():
    try: n=int(input('Number of queens (the slide exercise uses 4): ')); state=list(map(int,input('Enter one row per column, 0 through n-1: ').split()))
    except ValueError: print('Enter an integer size and integer rows.'); return
    if n<4 or len(state)!=n or sorted(state)!=list(range(n)):
        print('Use n >= 4 and a permutation of 0 through n-1.'); return
    result=hill_climb(state); print(f'\nFinal state: {result}; cost={cost(result)}')


SAMPLE_INPUTS = ['4', '0 1 2 3']
with sample_input(SAMPLE_INPUTS):
    main()
```
