## Program 1 — Tic-Tac-Toe with a minimax computer player

**Theory.** The deck describes a two-player game and introduces a game tree: each node is a board state and each edge is a legal move. Minimax searches terminal outcomes, choosing moves that maximize the human-side score and minimize it for the computer. The computer therefore chooses an optimal move for this small game.

**Objective.** Play against a computer that evaluates all remaining Tic-Tac-Toe moves.

**Libraries.** Python standard library (`math`).

**Algorithm.**
1. Represent the board as nine cells. 2. Check rows, columns and diagonals for a winner. 3. Recursively try each legal move. 4. Score a win, loss or draw. 5. Choose the computer move with the best minimax score.

**Complexity.** At most O(9!) game-tree nodes; O(9) recursion depth, excluding the board-state search tree.

The code below keeps `input()` so you can provide your own values. The final lines run a sample interaction through the same prompts; edit `SAMPLE_INPUTS` to try another case, or call the main function directly for a live terminal run.

**Expected output and explanation.** Minimax explores the complete remaining tree; depth is bounded by nine moves.


```python
from contextlib import contextmanager
import builtins

@contextmanager
def sample_input(values):
    original_input = builtins.input
    remaining = iter(map(str, values))
    def demo_input(prompt=''):
        print(prompt, end='')
        try:
            value = next(remaining)
        except StopIteration as exc:
            raise RuntimeError('Add more values to SAMPLE_INPUTS for this run.') from exc
        print(value)
        return value
    builtins.input = demo_input
    try:
        yield
    finally:
        builtins.input = original_input

import math

WIN_LINES = ((0,1,2),(3,4,5),(6,7,8),(0,3,6),(1,4,7),(2,5,8),(0,4,8),(2,4,6))

def winner(board):
    for a,b,c in WIN_LINES:
        if board[a] != ' ' and board[a] == board[b] == board[c]:
            return board[a]
    return None

def show_board(board):
    print('\n' + '\n---+---+---\n'.join(' | '.join(board[i:i+3]) for i in (0,3,6)))

def minimax(board, player):
    result = winner(board)
    if result == 'O': return 1
    if result == 'X': return -1
    if ' ' not in board: return 0
    scores=[]
    for i,cell in enumerate(board):
        if cell == ' ':
            board[i]=player
            scores.append(minimax(board, 'X' if player=='O' else 'O'))
            board[i]=' '
    return (max if player=='O' else min)(scores)

def computer_move(board):
    best_score=-math.inf
    best_index=None
    for i,cell in enumerate(board):
        if cell==' ':
            board[i]='O'
            score=minimax(board,'X')
            board[i]=' '
            if score>best_score:
                best_score,best_index=score,i
    return best_index

def main():
    board=[' ']*9
    print('You are X; the computer is O. Enter a square from 1 to 9.')
    while True:
        show_board(board)
        if winner(board):
            break
        if ' ' not in board:
            break
        try:
            pos=int(input('Your move (1-9): '))-1
        except ValueError:
            print('Please enter an integer from 1 to 9.'); continue
        if not 0<=pos<9 or board[pos]!=' ':
            print('That square is unavailable. Try again.'); continue
        board[pos]='X'
        if winner(board) or ' ' not in board:
            break
        pos=computer_move(board)
        board[pos]='O'
        print(f'Computer chooses square {pos+1}.')
    show_board(board)
    result=winner(board)
    print(f'{result} wins!' if result else 'The game is a draw.')

SAMPLE_INPUTS = ['1', '2', '3', '4', '5', '6', '7', '8', '9', '1', '2', '3', '4', '5', '6', '7', '8', '9', '1', '2', '3', '4', '5', '6', '7', '8', '9', '1', '2', '3', '4', '5', '6', '7', '8', '9', '1', '2', '3', '4', '5', '6', '7', '8', '9']
with sample_input(SAMPLE_INPUTS):
    main()
```

## Program 2 — Two-room vacuum agents

**Theory.** The vacuum-world model has two locations, A and B; each can be clean or dirty. A simple reflex agent uses the current location and percept. A goal-based agent also compares the current state to a target. These are small examples of an agent mapping percepts to actions.

**Objective.** Compare a simple reflex policy with a goal-directed policy using user-supplied room states.

**Libraries.** Python standard library only.

**Algorithm.**
1. Read the starting room and both room states. 2. If the current room is dirty, suck. 3. Otherwise move to the other room. 4. Stop the reflex run when both rooms are clean. 5. Run the goal-based policy toward a selected target room.

**Complexity.** O(1) time and space for this fixed two-room world.

The code below keeps `input()` so you can provide your own values. The final lines run a sample interaction through the same prompts; edit `SAMPLE_INPUTS` to try another case, or call the main function directly for a live terminal run.

**Expected output and explanation.** The reflex run cleans both rooms; the second run demonstrates reaching the chosen target room.


```python
from contextlib import contextmanager
import builtins

@contextmanager
def sample_input(values):
    original_input = builtins.input
    remaining = iter(map(str, values))
    def demo_input(prompt=''):
        print(prompt, end='')
        try:
            value = next(remaining)
        except StopIteration as exc:
            raise RuntimeError('Add more values to SAMPLE_INPUTS for this run.') from exc
        print(value)
        return value
    builtins.input = demo_input
    try:
        yield
    finally:
        builtins.input = original_input

def reflex_action(location, status):
    return 'Suck' if status=='Dirty' else ('Move Right' if location=='A' else 'Move Left')

def goal_action(location, status, goal):
    if status=='Dirty': return 'Suck'
    if location==goal: return 'Goal Reached'
    return 'Move Right' if location=='A' else 'Move Left'

def main():
    location=input('Starting location (A/B): ').strip().upper()
    if location not in {'A','B'}:
        print('Location must be A or B.'); return
    rooms={}
    for room in ('A','B'):
        status=input(f'Status of room {room} (Clean/Dirty): ').strip().capitalize()
        if status not in {'Clean','Dirty'}:
            print('Status must be Clean or Dirty.'); return
        rooms[room]=status
    print('\n--- Simple Reflex Agent ---')
    current=location
    for _ in range(4):
        action=reflex_action(current,rooms[current])
        print(f'Location: {current}  Status: {rooms[current]}  Action: {action}')
        if action=='Suck': rooms[current]='Clean'
        elif action=='Move Right': current='B'
        elif action=='Move Left': current='A'
        if all(value=='Clean' for value in rooms.values()): break
    print('Both rooms are clean!' if all(value=='Clean' for value in rooms.values()) else 'Cleaning stopped.')
    goal=input('Goal room for goal-based agent (A/B): ').strip().upper()
    if goal not in {'A','B'}: print('Goal must be A or B.'); return
    current=location
    print('\n--- Goal-Based Agent ---')
    for _ in range(4):
        action=goal_action(current,rooms[current],goal)
        print(f'Location: {current}  Status: {rooms[current]}  Action: {action}')
        if action=='Goal Reached': break
        if action=='Suck': rooms[current]='Clean'
        elif action=='Move Right': current='B'
        elif action=='Move Left': current='A'



SAMPLE_INPUTS = ['A', 'Dirty', 'Dirty', 'B']
with sample_input(SAMPLE_INPUTS):
    main()
```
